## 1. Understanding the Dataset

### 1.1 Loading and Initial Inspection

The dataset consists of two separate CSV files provided in the supplementary material of Antonio et al. (2019):
- `H1.csv`: booking records for a **Resort Hotel** in the Algarve region, Portugal.
- `H2.csv`: booking records for a **City Hotel** in Lisbon, Portugal.

Both tables share the same 31 columns. We load both files from `../data/`, add a categorical column `Hotel` to identify the property type (`Resort Hotel` vs. `City Hotel`), and concatenate them into a single DataFrame to inspect the raw structure before any cleaning.

We begin by importing the libraries used throughout the analysis.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import statsmodels.api as sm
import statsmodels.formula.api as smf

pd.set_option("display.max_columns", None)
```

We now load the two raw CSV files and check the size of each source table before adding the `Hotel` column and concatenating them into a single raw DataFrame.

```python
h1_raw = pd.read_csv("../data/H1.csv")
h2_raw = pd.read_csv("../data/H2.csv")

# Size of each source table before adding the Hotel column
print(f"H1 (Resort Hotel):      {h1_raw.shape[0]:>7,} rows x {h1_raw.shape[1]} columns")
print(f"H2 (City Hotel):        {h2_raw.shape[0]:>7,} rows x {h2_raw.shape[1]} columns")

h1_raw["Hotel"] = "Resort Hotel"
h2_raw["Hotel"] = "City Hotel"

df_raw = pd.concat([h1_raw, h2_raw], ignore_index=True)
print(f"Combined (with Hotel):  {df_raw.shape[0]:>7,} rows x {df_raw.shape[1]} columns")
```
